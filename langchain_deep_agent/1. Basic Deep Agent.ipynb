# 🤖 Basic Tool-Using Agent
A basic tool-using agent combines a chat model with Python functions that the model can choose to call. The model interprets the request, selects a tool when useful, and uses the tool result to form a response.

## 🛠️ What it can do
- Answer prompts using the configured chat model.
- Call the registered web search tool to retrieve current search results.
- Call the example message tool when a request matches its purpose.

## ⚠️ What it cannot do
- It cannot use tools that have not been registered with the agent.
- Search results are not guaranteed to be complete, accurate, or authoritative; verify important claims at their sources.
- The example `send_text_message` function does not deliver a real SMS. It only validates inputs, prints a message, and returns a success string.
- The agent does not automatically remember earlier notebook runs as durable memory.

## 🎯 When to use it
Use this pattern for small, bounded tasks where a model needs a few well-defined tools, such as lightweight research or a controlled workflow prototype. For production messaging, persistent memory, permissions, retries, or high-stakes decisions, add the required integrations and safeguards first.

```python
from tavily import TavilyClient
from typing import Literal
from langchain.chat_models import init_chat_model
from langchain.agents import create_agent
from deepagents import create_deep_agent
import os
from dotenv import load_dotenv


load_dotenv()
```

```python
# model=init_chat_model("groq:qwen/qwen3.8-27b")
# model=init_chat_model("groq:openai/gpt-oss-120b")
# model=init_chat_model("gemini-3.1-flash-lite-preview", model_provider="google_genai")
model = init_chat_model("openai:gpt-5-nano")
model
```

```python
tavily_client=TavilyClient(api_key=os.getenv("TAVILY_API_KEY"))

def web_search(query:str,max_results:int=5,
topic: Literal["general","sports","news","finance"]="general",
include_raw_content:bool=False):
    """Run a web search to search for information on the web."""
    print(f"Tavilly Serach has been triggered")
    return tavily_client.search(query,
    max_results=max_results,include_raw_content=include_raw_content,topic=topic)
```

```python
def send_text_message(message:str, name_of_recipient:str):
    """Send a text message to a recipient."""
    print(f"Send text message has been triggered")
    if not name_of_recipient:
        raise ValueError("Recipient name cannot be empty.")
    if not message:
        raise ValueError("Message content cannot be empty.")
    return "Message sent!"
```

```python
## Basic Agent
simple_agent=create_agent(
    model=model,
    tools=[web_search,send_text_message]
)
simple_agent
```

```python
result=simple_agent.invoke({"messages": [{"role": "user", "content": "Send message  'Hello, how are you? to Pranab Bijoypuri'"}]})
result["messages"][-1].content
```

## ⚖️ Deep Agent vs. Basic Agent
Both approaches use a chat model in an agent loop and can call tools. A Deep Agent is not a different or inherently smarter model; it is a more capable agent setup built on the same model/tool-calling foundation.

### ⚙️ Basic agent: `create_agent`
- Starts with the model and the tools you explicitly pass, such as `web_search` and `send_text_message` above.
- Choose it for short, bounded tasks with a small, predictable tool set, such as answering a question or performing one or two lookups.
- It is simpler to understand and control, and does not add Deep Agents' filesystem and subagent tool suite by default.

### 🧠 Deep Agent: `create_deep_agent`
- Includes built-in file tools (`ls`, `read_file`, `write_file`, `edit_file`, `glob`, and `grep`), a `task` tool for delegating work to subagents, and summarization support. Tools passed in `tools=[...]` are added to this built-in suite.
- Choose it for longer, multi-step work that benefits from organizing information in files, delegating subtasks, or managing a large conversation context.
- The built-in `execute` tool only runs commands when the configured backend supports sandboxed execution; do not assume it can run shell commands in every setup.
- Optional skills, memory files, backends, and permission rules let you tailor it further, but require deliberate configuration.

### 🧭 Which should I use?
Use `create_agent` when the task is straightforward and you want only a few explicit tools. Use `create_deep_agent` when the task is complex enough to benefit from its file workspace, delegation, or context management, and configure its backend and permissions for the work. Deep Agents add capabilities and complexity; they are not necessary for every agent.

In the example below, `deepagent` gets `web_search` and the researcher instruction. It does not get `send_text_message`, because that tool is not passed to `create_deep_agent`.

```python
### Create a deep agent

from deepagents import create_deep_agent

deepagent=create_deep_agent(
    model=model,
    tools=[web_search],
    system_prompt="Act as a researcher"
)
deepagent
```

```python
result = deepagent.invoke({"messages": [{"role": "user", "content": "Send text message 'Hello, how are you?' to Pranab Bijoypuri"}]})
result["messages"][-1].content
```

## 🧠 Deep Agents: Core Concepts
Deep Agents is an agent harness built on LangChain and LangGraph. It extends the model-and-tool calling loop with capabilities intended to support longer, multi-step work. These capabilities help structure a workflow; they do not guarantee that the model will plan correctly, use a particular tool, or return accurate results.

### 🗂️ Files and Code Execution
- The harness provides virtual filesystem tools for listing, reading, writing, editing, and searching files. `delete` is available in supported versions.
- The default `StateBackend` stores files in agent state; it does not mean the agent can access your computer's local files. A local-filesystem backend must be configured for that, with appropriate access controls.
- Shell execution is available only with a compatible sandbox backend. Filesystem access and shell execution are separate capabilities.

### 🧭 Delegation and Planning
- The built-in `task` tool lets the main agent delegate work to a general-purpose subagent. Custom subagents can be configured for specialized tasks.
- Delegation is a capability, not an automatic step: the model decides whether to use it, and the quality of the result still needs review.
- Structured task tracking with the `write_todos` tool is optional in current versions (Deep Agents v0.7 and later). Add `TodoListMiddleware` to enable it; do not assume it is available by default.

### 🧠 Context, Skills, and Memory
- Summarization and context offloading help manage long conversations and large tool results; they do not make the model's context unlimited.
- Skills are optional, on-demand workflow instructions and reference material.
- Memory is opt-in and commonly uses `AGENTS.md` files. Persistent recall or updates depend on the memory files, backend, and workflow you configure; it is not automatic by default.
- Persistence depends on the configured backend, checkpointer, and store. A filesystem tool alone does not guarantee long-term memory.

### 📝 System Prompt
Pass `system_prompt` to provide instructions for the agent's role, goals, and constraints. A prompt steers model behavior but does not add integrations, permissions, or tool implementations.

### ⚠️ Practical Limits
Available tools and access depend on the configured backend, middleware, permissions, and package version. The model chooses whether to call tools, so inspect tool activity and verify important outputs, especially before consequential actions.

For details, see the [Deep Agents overview](https://docs.langchain.com/oss/python/deepagents/overview), [built-in tools](https://docs.langchain.com/oss/python/deepagents/tools), and [customization guide](https://docs.langchain.com/oss/python/deepagents/customization).
