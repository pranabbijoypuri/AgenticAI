### Evaluator-optimizer Design Pattern

 -  Details are present in the **5-Evaluator-optimizer Architecture.md** file

```python
import os
from dotenv import load_dotenv
from langchain_groq import ChatGroq
from langchain_openai import ChatOpenAI
load_dotenv()


MODEL_NAME = os.getenv("MODEL_NAME")

# llm=ChatGroq(model="qwen/qwen3.8-27b",max_tokens=100,temperature=0.7)
llm = ChatOpenAI(model=MODEL_NAME)
```

```python
from typing import Annotated, List
import operator
from typing_extensions import Literal
from pydantic import BaseModel,Field
from langchain_core.messages import HumanMessage,SystemMessage
from typing_extensions import TypedDict
```

```python
# Graph state
class State(TypedDict):
    joke: str
    topic: str
    feedback: str
    funny_or_not: str
    iteration: int # The current iteration of joke generation and evaluation
    max_iterations: int # The maximum number of iterations allowed for joke generation and evaluation

# Schema for structured output to use in evaluation
class Feedback(BaseModel):
    grade: Literal["funny", "not funny"] = Field(
        description="Decide if the joke is funny or not.",
    )
    feedback: str = Field(
        description="If the joke is not funny, provide feedback on how to improve it.",
    )

# Augment the LLM with schema for structured output
evaluator = llm.with_structured_output(Feedback)
```

```python
# This function is called by the orchestrator to generate a joke based on the topic and feedback (if any)
def llm_call_generator(state: State):
    """LLM generates a joke"""

    print(f" Generating joke for topic.")
    if state.get("feedback"):
        msg = llm.invoke(
            f"Write a joke about {state['topic']} but take into account the feedback: {state['feedback']}"
        )
    else:
        msg = llm.invoke(f"Write a joke about {state['topic']}")

    return {
                "joke": msg.content, 
                "iteration": state.get("iteration", 0) + 1
            }

# This function is called by the orchestrator to evaluate the joke generated by the LLM
def llm_call_evaluator(state: State):
    """LLM evaluates the joke"""
    print(f" Evaluating joke...") 
    grade = evaluator.invoke(f"Grade the joke {state['joke']}")
    print(f" Joke evaluation: {grade.grade}, Feedback: {grade.feedback}")
    return {"funny_or_not": grade.grade, "feedback": grade.feedback}

# Conditional edge function to route back to joke generator or end based upon feedback from the evaluator. 
# The function checks if the joke was funny or not and if the maximum number of iterations has been reached. If the joke is funny, it returns "Accepted". 
# If the joke is not funny and the maximum number of iterations has not been reached, it returns "Rejected + Feedback". 
# If the joke is not funny and the maximum number of iterations has been reached, it returns "Accepted".

def route_joke(state: State):
    """Route back to joke generator or end based upon feedback from the evaluator"""

    if state["funny_or_not"] == "funny":
        return "Accepted"
    elif state["funny_or_not"] == "not funny" and state["iteration"] < state["max_iterations"]:
        return "Rejected + Feedback"
    else:
        print(f" Maximum iterations reached. Ending joke generation.")
        return "Accepted"
```

```python
from langgraph.graph import StateGraph, START, END
from IPython.display import Image, display
```

```python
# Build workflow
optimizer_builder = StateGraph(State)

# Add the nodes
optimizer_builder.add_node("llm_call_generator", llm_call_generator)
optimizer_builder.add_node("llm_call_evaluator", llm_call_evaluator)

# Add edges to connect nodes
optimizer_builder.add_edge(START, "llm_call_generator")
optimizer_builder.add_edge("llm_call_generator", "llm_call_evaluator")
optimizer_builder.add_conditional_edges(
    "llm_call_evaluator",
    route_joke,
    {  # Name returned by route_joke : Name of next node to visit
        "Accepted": END,
        "Rejected + Feedback": "llm_call_generator",
    },
)

# Compile the workflow
optimizer_workflow = optimizer_builder.compile()
# Show the workflow
display(Image(optimizer_workflow.get_graph().draw_mermaid_png()))
```

```python
# Invoke
state = optimizer_workflow.invoke({"topic": "Agentic AI system", "max_iterations": 3})
print(state["joke"])
```
