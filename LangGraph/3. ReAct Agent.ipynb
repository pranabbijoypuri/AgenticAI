```python
import os
from langchain_google_genai import ChatGoogleGenerativeAI
from langgraph.prebuilt import create_react_agent
from langchain_community.tools.tavily_search import TavilySearchResults

# 1. Setup the Google Model
# Ensure GOOGLE_API_KEY is in your environment
llm = ChatGoogleGenerativeAI(model="gemini-1.5-flash", temperature=0)

# 2. Define tools
tools = [TavilySearchResults(max_results=1)]

# 3. Create the agent
# In LangGraph, 'state_modifier' replaces the need for the hub prompt
agent = create_react_agent(
    llm,
    tools,
    state_modifier="You are a helpful assistant. Use tools to verify facts."
)

# 4. Invoke the agent
inputs = {"messages": [("user", "Who won the best picture Oscar in 2024?")]}
for output in agent.stream(inputs, stream_mode="updates"):
    print(output)
```
